```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Data

ns_5_1 = [2, 17, 1, 346]
ns_5_2 = [5, 24, 8, 51]
ss_30_1 = [53, 94, 14, 191]
ss_30_2 = [27, 87, 15, 168]
ss_30_3 = [31, 76, 24, 158]
v1_30 = [19, 88, 19, 124]
v2_30 = [24, 60, 13, 121]
v3_30_1 = [25, 69, 16, 129]
v3_30_2 = [29, 63, 11, 106]
```

```python
def plot_rates(datasets, durations, sum_mask):
    counts = np.array([np.dot(data, sum_mask) for data in datasets])
    rates = counts / durations
    rel_uncertainties = 1 / np.sqrt(counts)
    uncertainties = rates * rel_uncertainties

    x_axis = range(len(datasets))
    plt.errorbar(x_axis, rates, yerr=uncertainties)


plot_rates([ss_30_1, v1_30, v2_30, v3_30_1], [30, 30, 30, 30], [0, 0, 0, 1])
```
