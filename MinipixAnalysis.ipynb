```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Data

ns_5_1 = np.array([2, 17, 1, 346])
ns_5_2 = np.array([5, 24, 8, 51])
ss_30_1 = np.array([53, 94, 14, 191])
ss_30_2 = np.array([27, 87, 15, 168])
ss_30_3 = np.array([31, 76, 24, 158])
v1_30 = np.array([19, 88, 19, 124])
v2_30 = np.array([24, 60, 13, 121])
v3_30_1 = np.array([25, 69, 16, 129])
v3_30_2 = np.array([29, 63, 11, 106])

ss_90_all = ss_30_1 + ss_30_2 + ss_30_3
v3_60_all = v3_30_1 + v3_30_2
ns_10_all = ns_5_1 + ns_5_2

BLOCK_HEIGHT = 4.7
```

```python
# data sets = array containing trials to use
# durations = corresponding times over which data was taken
# sum_mask = array encoding which particles to consider, 1 for true, 0 for false
    # [alphas, electrons, muons, dots ]
def plot_rates(datasets, durations, sum_mask, title="Rate vs Shielding Intensity", color="r"):
    counts = np.array([np.dot(data, sum_mask) for data in datasets])
    rates = counts / durations
    rel_uncertainties = 1 / np.sqrt(counts)
    uncertainties = rates * rel_uncertainties

    x_axis = np.array(range(len(datasets))) * BLOCK_HEIGHT
    plt.errorbar(x_axis, rates, yerr=uncertainties, 
                 fmt="s", c=color, ecolor='k', capsize=5, linestyle="", barsabove=True)
    plt.title(title)
    plt.ylabel("Rate (Counts per minute)")
    plt.xlabel("Shielding amount (vertical cm of lead)")
    plt.xlim([-0.5, 20])


#plot_rates([ss_90_all, v1_30, v2_30, v3_60_all], [90, 30, 30, 60], [0, 1, 1, 0], title="All Data")
plot_rates([ss_30_1, v1_30, v2_30, v3_30_1], [30, 30, 30, 30], [0, 1, 1, 0], title="First Run", color="b")

#plot_rates([ns_10_all, ss_90_all, v1_30, v2_30, v3_60_all], [10, 90, 30, 30, 60], [1, 0, 0, 0]) # Doesnt really make sense to plot the ns run
                                                                                                 # on the same scale as the others but oh well
```
